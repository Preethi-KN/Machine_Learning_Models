```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_iris
```

```python
iris=load_iris()
X=iris.data
y=iris.target
y
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test=train_test_split(X,y, test_size=0.2)
```

```python
from sklearn.tree import DecisionTreeClassifier
clf=DecisionTreeClassifier()
clf.fit(X_train,y_train)
y_pred=clf.predict(X_test)
```

```python

from sklearn.metrics import accuracy_score
accuracy_score(y_test,y_pred)
```

```python
from sklearn.tree import plot_tree
from matplotlib.pylab import rcParams
rcParams['figure.figsize'] = 80,50

plot_tree(clf)
```

```python
import matplotlib.pyplot as plt 

def analyser(max_depth):
    
    X=data.iloc[:, 2:4].values
    y=data.iloc[:,-1].values
    #set max depth 
    clf2 = DecisionTreeClassifier(max_depth=max_depth)
    clf2.fit(X, y)

    # 2. Set identical mesh steps (0.1) for smooth boundary plotting
    a = np.arange(start=X[:, 0].min() - 1, stop=X[:, 0].max() + 1, step=0.1)
    b = np.arange(start=X[:, 1].min() - 1, stop=X[:, 1].max() + 1, step=100)

    XX, YY = np.meshgrid(a, b)

    input_arry = np.array([XX.ravel(), YY.ravel()]).T

    # 3. Use 'clf2' to predict, matching your trained model name
    lables = clf2.predict(input_arry)

    # 4. Fixed the 'contourf' typo
    plt.contourf(XX, YY, lables.reshape(XX.shape), alpha=0.5)
    plt.scatter(X[:, 0], X[:, 1], c=y,s=1000, edgecolor='k')
    plt.show()
```

```python
analyser(max_depth=None)  # This will default to max_depth=None automatically -> overfitting
#OR
#analyser(max_depth=1) # underfit
# OR
#analyser(max_depth=5)
```
