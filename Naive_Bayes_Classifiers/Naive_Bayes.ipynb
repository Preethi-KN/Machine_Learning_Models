# Problem 1

# Outlook=Sunny, Temp=Hot, Humidity=High, Wind=Weak
# Play or No Play?

# Solution

# P(Yes | Sunny,Hot,High,Weak)=P(Sunny | Yes) * P(Hot | Yes) * P(High | Yes) * P(Weak | Yes) *P(Yes)

# P(No | Sunny,Hot,High,Weak)=P(Sunny | No) * P(Hot | No) * P(High | No) * P(Weak | No)*P(No)

# Compare and decide using the Maximum a posteriori rule

```python
#training
#data is stored in the lookup table
#P(Yes) and #P(No)
df['play'].value_counts()
```

```python
py=9/14
pn=5/14
```

```python
print("P(yes):",py)
print("P(No):",pn)
```

```python
#Outlook
pd.crosstab(df['outlook'],df['play'])
```

```python
# prob  p(n,x1,x2,x3)
p_overcast_n=0
p_rain_n=2/5
p_suuny_n=3/5

# prob  p(y,x1,x2,x3)
p_overcast_y=4/9
p_rain_y=3/9
p_suuny_y=2/9
```

```python
#Temp
pd.crosstab(df['temp'],df['play'])
```

```python
# prob No  p(n,x1,x2,x3)
p_cool_n=1/5
p_hot_n=2/5
p_mild_n=2/5

# prob yes  p(n,x1,x2,x3)
p_cool_y=3/9
p_hot_y=2/9
p_mild_y=4/9
```

```python
#humidity
pd.crosstab(df['humidity'],df['play'])
```

```python
# prob No  p(n,x1,x2,x3)
p_high_n=4/5
p_normal_n=1/5

# prob yes  p(n,x1,x2,x3)
p_high_y=3/9
p_normal_y=6/9
```

```python
#wind
pd.crosstab(df['wind'],df['play'])
```

```python
# prob No  p(n,x1,x2,x3)
p_strong_n=3/5
p_weak_n=2/5

# prob yes  p(n,x1,x2,x3)
p_strong_y=3/9
p_weak_y=6/9
```

```python
# Outlook=Sunny, Temp=Hot, Humidity=High, Wind=Weak
# Play or No Play?
p_yes=py*p_suuny_y*p_hot_y*p_high_y*p_weak_y
print(p_yes)
```

```python
p_No=pn*p_suuny_n*p_hot_n*p_high_n*p_weak_n
print(p_No)
```
